```python
# Confirm which columns have nulls and how many
df.isna().sum()
```

```python
# Fill Embarked with its most common value. Find that value with code, don't hardcode it (.mode())
most_common = df["Embarked"].mode()[0]
df["Embarked"] = df["Embarked"].fillna(most_common)
```

```python
# Compare Age mean vs median. Which would you fill with and why?
print(df["Age"].mean())
print(df["Age"].median())
# I would fill it with median as mean is affected by very large or very small values
```

```python
# Fill Age with the overall median. Check the mean and std before and after — what moved?
# Mean and standard deviation before filling
print("Before:")
print("Mean:", df["Age"].mean())
print("Std:", df["Age"].std())

# Fill missing ages with the median
median = df["Age"].median()
df["Age"] = df["Age"].fillna(median)

# Mean and standard deviation after filling
print("After:")
print("Mean:", df["Age"].mean())
print("Std:", df["Age"].std())
# THe Mean slightly changes but the std deviatio changed a bit dont know why
```

```python
# Now do it by class instead, using the transform line. Compare the resulting mean to Q4's.
print("Before:")
print("Mean:", df["Age"].mean())
df["Age"] = df.groupby("Pclass")["Age"].transform(lambda x: x.fillna(x.median()))
print("After:")
print("Mean:", df["Age"].mean())
```

```python
# Create a new column HasCabin — 1 if Cabin is recorded, 0 if not. Then compare survival rate by HasCabin. What do you find, and why?
df["HasCabin"] = df["Cabin"].notna().astype(int)
print(df.groupby("HasCabin")["Survived"].mean())
# Passenger's with a cabin has more survival rate due to higher access of lifboat
```

```python
import pandas as pd
df = pd.read_csv("titanic.csv")
df.shape
```

```python
df_overall = df.copy()
df_overall["Age"] = df_overall["Age"].fillna(df_overall["Age"].median())

df_byclass = df.copy()
df_byclass["Age"] = df_byclass.groupby("Pclass")["Age"].transform(lambda x: x.fillna(x.median()))

print("Overall:", df_overall["Age"].mean(), df_overall["Age"].std())
print("By class:", df_byclass["Age"].mean(), df_byclass["Age"].std())
```
