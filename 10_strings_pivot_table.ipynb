```python
import pandas as pd

titanic = pd.read_csv("titanic.csv")
df = pd.read_csv("titanic.csv")

titanic.shape
```

```python
titanic["Name"].head()
```

```python
titanic["Name"].str.split(",").head()
```

```python
titanic["Title"] = titanic["Name"].str.split(",").str[1].str.split(".").str[0].str.strip()
titanic["Title"].value_counts()
```

```python
titanic.groupby("Title")["Survived"].agg(["mean", "count"])
```

```python
titanic["Title"] = titanic["Title"].replace({"Mlle": "Miss", "Ms": "Miss", "Mme": "Mrs"})

common = ["Mr", "Miss", "Mrs", "Master"]
titanic["Title"] = titanic["Title"].where(titanic["Title"].isin(common), "other")

titanic.groupby("Title")["Survived"].agg(["mean", "count"])
```

```python
df.groupby(["Pclass", "Sex"])["Survived"].mean().unstack()          # Lesson 3
```

```python
pd.crosstab(df["Pclass"], df["Sex"], values=df["Survived"], aggfunc="mean")   # Lesson 8
```

```python
df.pivot_table(index="Pclass", columns="Sex", values="Survived", aggfunc="mean")
```

```python
titanic.pivot_table(index="Pclass", columns="Sex", values="Survived",
                    aggfunc=["mean", "count"])
```

```python
titanic.pivot_table(index="Pclass", values=["Survived", "Fare", "Age"],
                    aggfunc="mean")
```

```python
titanic.pivot_table(index="Pclass",
                    values=["Survived", "Fare"],
                    aggfunc={"Survived": "mean", "Fare": "median"})
```
