```python
import pandas as pd

df = pd.read_csv("titanic.csv")
df.shape
```

```python
# Average fare by passenger class
print(df.groupby("Pclass")["Fare"].mean())
```

```python
# Survival rate by Embarked port
print(df.groupby("Embarked")["Survived"].mean())
```

```python
# Count of passengers by class and sex together
print(df.groupby(["Pclass","Sex"]).size())
```

```python
# Average age by class — is there a pattern, and what explains it?
print(df.groupby("Pclass")["Age"].mean())
# Yes, first-class passengers were older on average, likely because wealthier and older people tended to travel in first class.
```

```python
# Which class had highest survival rate, which lowest?
print(df.groupby("Pclass")["Survived"].mean())
```
