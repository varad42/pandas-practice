```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("temps.csv")
df["Date"] = pd.to_datetime(df["Date"])
df_t = df.set_index("Date")

titanic = pd.read_csv("titanic.csv")

df_t.shape, titanic.shape
```

```python
df_t["Temp"].plot()
```

```python
df_t["Temp"].plot(figsize=(12, 4))
```

```python
df_t["Temp"].rolling(30).mean().plot(figsize=(12, 4))
```

```python
df_t["Temp"].plot(kind="hist", bins=30)
```

```python
df_t.groupby(df_t.index.month)["Temp"].mean().plot(kind="bar")
```

```python
titanic = pd.read_csv("titanic.csv")
titanic.plot(kind="scatter", x="Age", y="Fare")
```

```python
titanic.plot(kind="scatter", x="Age", y="Fare", ylim=(0, 150), alpha=0.3)
```

```python
# Histogram of Fare. Compare its shape to the Temp histogram — what's different, and what does that tell you about fares?

titanic["Fare"].plot(kind="hist", bins=30)

# fare histogram shows that fare are unevenly spread and temp hist is more concentrated
```

```python
# Bar chart of survival rate by Pclass.
titanic.groupby("Pclass")["Survived"].mean().plot(kind="bar")
```

```python
# Bar chart of average Temp by day of week. Predict the shape before plotting, and say why.
# it willl be flat for all days

df_t.groupby(df_t.index.day_name())["Temp"].mean().plot(kind="bar")
```
