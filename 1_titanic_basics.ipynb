```python
mask = df["Age"] > 30
mask                              # look at it
mask.sum()                        # why does summing a boolean column work?

df[df["Age"] > 30]
df[df["Sex"] == "female"]
df[(df["Age"] > 30) & (df["Sex"] == "female")]
df[(df["Pclass"] == 1) | (df["Pclass"] == 2)]
df[df["Pclass"].isin([1, 2])]     # cleaner version of the line above
df[df["Age"].isna()]              # the missing-age rows
df[~(df["Sex"] == "male")]        # ~ means NOT
```

```python
# How many passengers were in 1st class?
len(df[df["Pclass"] == 1])
```

```python
# How many women survived?
len(df[(df["Survived"] == 1) & (df["Sex"] == "female")])
```

```python
# What was the average fare paid by 1st class passengers? (Filter, then take the mean of Fare)
fare = df[df["Pclass"] == 1]["Fare"].mean()
print(fare)
```

```python
# How many passengers under 18 were on board?
count = df[df["Age"] < 18]
len(count)
    
```

```python
# How many passengers paid a fare of exactly 0?
zero = df[df["Fare"] == 0]
print(zero)
```

```python
# Sort the dataframe by Fare descending and show the top 5 — who paid the most?
df.sort_values("Fare", ascending=False).head()
```

```python
# Compare survival rate for men vs women. Use the mean-of-binary trick from Q6.
male_rate = df[df["Sex"] == "male"]["Survived"].mean()
female_rate = df[df["Sex"] == "female"]["Survived"].mean()

print("Male survival rate:", male_rate)
print("Female survival rate:", female_rate)
```
