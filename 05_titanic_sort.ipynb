```python
def age_group(age):
    if pd.isna(age):
        return "Unknown"
    elif age < 18:
        return "Child"
    elif age < 60:
        return "Adult"
    else:
        return "Senior"

age_group(22)    # "Adult"
age_group(4)     # "Child"
age_group(71)    # "Senior"
```

```python
df["Age"].apply(age_group)
```

```python
df["AgeGroup"] = df["Age"].apply(age_group)
df["AgeGroup"].value_counts()
```

```python
df.groupby("AgeGroup")["Survived"].mean()
```
