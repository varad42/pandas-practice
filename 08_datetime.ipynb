```python
import pandas as pd

url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/daily-min-temperatures.csv"
df = pd.read_csv(url)
df.to_csv("temps.csv", index=False)

df.head()
df.shape
```

```python
df.dtypes
```

```python
df["Date"] = pd.to_datetime(df["Date"])
df.dtypes
```

```python
df["Date"].dt.day_name().value_counts()
```

```python
full = pd.date_range(df["Date"].min(), df["Date"].max())
full.difference(df["Date"])
```

```python
df.groupby(df["Date"].dt.year)["Temp"].mean()
```
